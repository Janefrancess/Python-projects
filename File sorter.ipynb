# File Sorter Useful For File Organization in Web Development.


```python
import os, shutil
```

```python
path = r"C:/Users/Stanley Jayne/Desktop/SortedFiles/"
```

```python
file_name = os.listdir(path)
```

```python
folder_names = ['pages', 'css', 'js', 'images', 'videos', 'audio', 'text_files']

for loop in range(0,7):
    if not os.path.exists(path + folder_names[loop]):
      os.makedirs(path + folder_names[loop])


for file in file_name:
    if file.lower().endswith(('.html', '.htm')) and not os.path.exists(path + "pages/" + file):
        shutil.move(path + file, path + "pages/" + file)
    elif file.lower().endswith(('.css')) and not os.path.exists(path + "css/" + file):
        shutil.move(path + file, path + "css/" + file)
    elif file.lower().endswith(('.js', '.mjs')) and not os.path.exists(path + "js/" + file):
        shutil.move(path + file, path + "js/" + file)
    elif file.lower().endswith(('.jpg', '.jpeg', '.png', '.gif')) and not os.path.exists(path + "images/" + file):
        shutil.move(path + file, path + "images/" + file)
    elif file.lower().endswith(('.mp4', '.mkv', '.avi', '.mov', '.webm')) and not os.path.exists(path + "videos/" + file):
        shutil.move(path + file, path + "videos/" + file)
    elif file.lower().endswith(('.mp3', '.wav', '.aac', '.flac', '.m4a', '.ogg')) and not os.path.exists(path + "audio/" + file):
        shutil.move(path + file, path + "audio/" + file)
    elif file.lower().endswith(('.txt', '.docx', '.doc', '.pdf', '.rtf')) and not os.path.exists(path + "text_files/" + file):
        shutil.move(path + file, path + "text_files/" + file)
        
    
```
